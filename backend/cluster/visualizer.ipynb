```python
from pathlib import Path
import sys
import numpy as np
import umap
BACKEND_DIR = Path.cwd().parent
print(BACKEND_DIR)
sys.path.append(str(BACKEND_DIR))

from storage.sqlite_db import get_connection, get_text_embedding
```

```python
result = get_text_embedding()
print(result)
```

```python
embeddings = []
for row in result:
    blob_data = row[1]
    if blob_data:
        embedding_array = np.frombuffer(blob_data,dtype=np.float32)
        embeddings.append(embedding_array)
print(len(embeddings))
```

```python
reducer = umap.UMAP(n_components=3,random_state=42)
points = reducer.fit_transform(embeddings)
print(points)
```

```python
len(points)
```

```python
import plotly.express as px
import pandas as pd

df = pd.DataFrame({
    'x' : points[:,0],
    'y' : points[:,1],
    'z' : points[:,2],
    'id' : [row[0] for row in result]
})

fig = px.scatter_3d(
    df,x='x',y='y',z='z',hover_data=['id'],title='3D Embeddings'
)

fig.update_traces(marker=dict(size=4))
fig.show()
```
